```python
import cv2                            # camera + face detection
import os                             # folder/files han
import numpy as np                    # data ko array mei convert karne ke liye

# dataset folder
if not os.path.exists("dataset"):   Face_recognize,face_detection
    os.makedirs("dataset")

recognizer = cv2.face.LBPHFaceRecognizer_create()          #Ye model hai jo faces ko learn + recognize karta hai

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)                    #Ye face detect karta hai (rectangle banata hai)
  
# train function
def train_model():            #Ye function:images uthata hai,labels assign karta hai,model train karta hai
    faces = []
    labels = []
    label_map = {}
    label_id = 0

    for person in os.listdir("dataset"):            #Har folder = ek person
        label_map[label_id] = person       #label ko assign karga
        person_path = os.path.join("dataset", person)

        for img_name in os.listdir(person_path):
            img = cv2.imread(os.path.join(person_path, img_name), cv2.IMREAD_GRAYSCALE)         #Image ko grayscale me convert kiya ,👉 Model grayscale pe best kaam karta hai
            faces.append(img)
            labels.append(label_id)

        label_id += 1

    if len(faces) > 0:
        recognizer.train(faces, np.array(labels))            #Model faces ko yaad kar leta hai
        return label_map
    
    return {}

label_map = train_model()

cap = cv2.VideoCapture(0)                # camera start Webcam ON

print("Press 'r' = Register | 'q' = Exit")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces_detected = face_cascade.detectMultiScale(gray, 1.3, 5)    #Face ke coordinates milte hain

    for (x, y, w, h) in faces_detected:
        face = gray[y:y+h, x:x+w]

        if len(label_map) > 0:
            label, confidence = recognizer.predict(face)      #Model batata hai:kaun hai (label), kitna sure hai (confidence)
            name = label_map.get(label, "Unknown")

            if confidence > 100:     # Confidence high = doubt → Unknown
                name = "Unknown"
        else:
            name = "No Data"

        cv2.rectangle(frame, (x,y), (x+w,y+h), (0,255,0), 2)       # Face pe box + naam
        cv2.putText(frame, name, (x,y-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0,255,0), 2)

    cv2.imshow("Face System", frame)

    key = cv2.waitKey(1) & 0xFF

    # 🔥 REGISTER (no input)
    if key == ord('r'):
        name = "vinayak" 
        
        person_path = os.path.join("dataset", name)
        if not os.path.exists(person_path):
            os.makedirs(person_path)

        count = 0
        print("Registering...")

        while count < 20:
            ret, frame = cap.read()
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            faces = face_cascade.detectMultiScale(gray, 1.3, 5)

            for (x, y, w, h) in faces:
                face = gray[y:y+h, x:x+w]
                cv2.imwrite(f"{person_path}/{count}.jpg", face)
                count += 1

                cv2.rectangle(frame, (x,y), (x+w,y+h), (255,0,0), 2)
                cv2.imshow("Registering", frame)

            if cv2.waitKey(1) & 0xFF == 27:
                break

        print("✅ Face Registered")

        label_map = train_model()
        

    # EXIT
    elif key == ord('q'):
        print("❌ Exiting...")
        break

cap.release()
cv2.destroyAllWindows()
```
