# RNN (Recurrent Neural Network)
RNN (Recurrent Neural Network) ek special type ka neural network hai jo sequential data ko process karne ke liye design kiya gaya hai. Yah un situations mein kaam aata hai jahan data ka ek sequence ho, jaise time series data, speech recognition, aur language processing. Samajhte hain isko thoda detail mein:

# RNN ka Concept:
Traditional neural networks mein input aur output independent hote hain, lekin bahut si real-world problems mein humko previous inputs ka bhi dhyan rakhna padta hai. For example, agar aap ek sentence ka meaning samajhna chahte ho, toh har word ka context important hota hai. RNN ka kaam yeh hota hai ki yeh pichle steps se jo information mili hai, usko yaad rakh kar current input ko process karta hai.

RNN mein ek feedback loop hota hai jo allow karta hai ki previous information ko future ke predictions ke liye use kiya ja sake. Is feedback loop ke through network apne aapko ‘recurrent’ karta hai.

Kaise Kaam Karta hai RNN:
Input Layer: Sabse pehle, input layer hoti hai jo data ko RNN mein input karti hai. Yeh input ek sequence ke form mein hota hai.

Hidden Layer (Memory): Hidden layer mein ek "memory" hoti hai jo previous inputs ki information ko store karti hai. Yeh layer current input ko previous hidden state ke saath combine karti hai aur fir se hidden state ko update karti hai. Iska matlab yeh hai ki RNN apne pichle inputs ko yaad rakhne ki koshish karta hai.

Output Layer: Yeh final output generate karti hai jo ya to ek prediction ho sakta hai ya ek aur processed data.

ek aisi neural network architecture hai jo sequence data ya time-series data ke liye use hoti hai. Iska special feature hai memory, matlab yeh past information ko retain kar sakti hai aur current output mein use kar sakti hai. Yeh kaafi useful hota hai jab data sequential hota hai jaise time-series data, text data, speech recognition, language translation, etc.

# RNN Real-Life Example:
1Text Prediction: Jab aap apne phone pe text likhte ho, aur keyboard aapko agla word predict karta hai, to wo RNN use karta hai. Wo pichle words ko yaad rakhta hai aur unke basis pe next word predict karta hai.

Stock Price Prediction: RNN stock prices predict karne mein madad karta hai by looking at past price trends.

Language Translation: Machine translation mein bhi RNN ka use hota hai, jahan ek language ke sentence ko doosri language mein translate karte hain.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder
from tensorflow.keras import layers,models
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
from sklearn.model_selection import train_test_split
df = pd.read_csv(r"C:\Users\Lenovo\OneDrive\Desktop\gaurav\covid_toy.csv")
df["gender"] = df["gender"].map({"Male":1,"Female":2})
l = LabelEncoder()
df["cough"] = l.fit_transform(df["cough"])
df["city"] = l.fit_transform(df["city"])
df["has_covid"] =l.fit_transform(df["has_covid"])
df.fillna(df.mean(numeric_only= True),inplace=True)
```

```python
X  = df.drop("has_covid",axis=1)
y = df["has_covid"]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=20,random_state=42)
s = StandardScaler()
X_train = s.fit_transform(X_train)
X_test = s.transform(X_test)
```

```python
# Data ko reshape karna for RNN (e.g. treating each feature as a timestep)
X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)  # (samples, timesteps, features)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)

# RNN model create karna
model = models.Sequential()

# RNN layer add karna (SimpleRNN ya LSTM ya GRU bhi use kar sakte hain)
model.add(layers.SimpleRNN(32, activation='relu', input_shape=(X_train.shape[1], 1)))

# Fully connected layers add karna
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(32, activation='relu'))

# Output layer for binary classification (has_covid ya not covid)
model.add(layers.Dense(1, activation='sigmoid'))

# Model ko compile karna (binary_crossentropy for binary classification)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Model ko train karna
model.fit(X_train, y_train, epochs=20, batch_size=32, validation_split=0.2)

# Model ka summary dekhna
model.summary()

# Model ko test data pe evaluate karna
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f'Test Accuracy: {test_acc:.2f}')
```

```python
RNN Basics:1
Memory (Hidden State): Yeh RNN ko normal neural network se alag banata hai. Isme har step pe ek hidden state hoti hai jo pichle inputs ko yaad rakhti hai.

Sequential Data Handling: Har input pe output depend karta hai pichle inputs pe bhi, jo isse powerful banata hai for sequences.

RNN Structure:
RNN ka structure cyclic hota hai, jisme ek loop hota hai jo past information ko retain karta hai. Har time step pe hidden state update hoti hai aur wo aage ke calculations mein use hoti hai.

Ab hum Titanic dataset pe RNN ka code likhte hain, though Titanic dataset mein sequential data directly nahi hai. Hum yeh example ke liye use karenge jahan hum passenger ke features ko time-step ke roop mein treat karenge to demonstrate RNN usage.

Challenges in RNN:
Vanishing Gradient Problem: Jab hum RNN mein backpropagation karte hain, toh gradients kuch steps ke baad bahut chhote ho jate hain, jis wajah se network pichle steps ki information properly nahi seekh pata. Yeh issue long-term dependencies ke case mein hota hai.

Exploding Gradient Problem: Kabhi-kabhi gradients excessively badh jate hain, jo network ko unstable kar deta hai.

Variants of RNN:
LSTM (Long Short-Term Memory): LSTM RNN ka ek improved version hai jo long-term dependencies ko handle karne mein madad karta hai. Ismein special gates hoti hain jo decide karti hain ki kaunsi information important hai aur kis information ko discard karna hai.

GRU (Gated Recurrent Unit): Yeh bhi LSTM jaisa hi hota hai lekin thoda simplified version hai. Yeh computationally thoda faster hota hai.

Applications of RNN:
Speech Recognition: RNN speech ke sequences ko samajhkar text me convert karne me kaam aata hai.

Language Translation: RNN ek language ke sentences ko dusre language me translate kar sakta hai, kyunki yeh context ko samajhne ki kshamta rakhta hai.

Time Series Forecasting: Jaise stock market ka data, weather predictions etc., jisme time ke saath data change hota hai, RNN us data ko analyze karke future predictions kar sakta hai.
```
